# Chroma DB Key Concepts and Architecture

## Core Capabilities

ChromaDB is a vector database built specifically for retrieval tasks. It supports:

- **Embedding + metadata storage** — stores vector representations along with associated metadata
- **Vector search** — semantic similarity search using distance metrics (cosine, Euclidean, dot product)
- **Full-text search** — lexical/spelling-based search, separate from semantic search
- **Document storage** — stores the actual original text/documents, not just their embeddings
- **Metadata filtering** — narrow results using metadata (e.g., filter by date, category, author) alongside similarity search
- **Multi-modal retrieval** — can handle images, audio, and text in a unified way, not just text

## Deployment Modes

| Mode | How it works | When to use |
|---|---|---|
| **Client-server** | Chroma client connects over HTTP to a Chroma server running as a separate process (started via CLI or Docker) | Production setups, remote access, multiple clients connecting to one server |
| **Standalone (embedded)** | Client and server run in the same single process — no separate server needed | Local development, quick testing, notebooks — likely what you'll use first |

If you are a beginner then you'll almost certainly start in **standalone mode** — it's simpler, no server to manage.

## Architecture — The Workflow Phases

1. **Obtain embeddings (optional step)**
   - You can generate embeddings yourself beforehand (using a model you choose), **or**
   - Let ChromaDB handle embedding automatically in the background — you just hand it raw text

2. **Create a collection**
   - A collection is ChromaDB's equivalent of a **table** in a relational database — it's where your data lives, and you give it a name

3. **Store data into the collection**
   - If you precomputed embeddings yourself → you pass them in directly
   - If not → ChromaDB computes and stores the embeddings for you automatically

4. **Collection operations**
   - Standard management: delete, update, or rename collections

5. **Query and retrieve**
   - Query using text or a vector directly
   - ChromaDB automatically embeds your query text too (you don't need to manually embed your search query)
   - Results can be filtered further using metadata or document content, on top of similarity search

## Distance Metrics

- **Default: Euclidean (L2) distance**
- Also supports **cosine distance** and **dot product**
- You can (and for text/RAG work, probably should) explicitly set the metric to **cosine** when creating a collection, per what you learned earlier — Euclidean isn't usually the best default for text embeddings

## Performance

- Uses **HNSW (Hierarchical Navigable Small World)** — an ANN (approximate nearest neighbor) algorithm — to make similarity search fast even at scale, instead of brute-force comparing every vector
- Core engine written in **Rust**, giving roughly **3–5x speed improvement** over a pure-Python implementation for read/write operations

ChromaDB's internal engine (the part that does indexing, HNSW search, storage) is written in Rust for speed. But it's wrapped in a Python API that you interact with so you can just write the code in python and ChromaDB will handle the Rust part for speed.

## Clients & Integrations

- **Officially supported:** Python, JavaScript (maintained by the ChromaCore team)
- **Community-supported (unofficial):** Ruby, Java, Go, C#, Rust, PHP
- **Framework integrations:** LangChain, LlamaIndex, Ollama — directly relevant since these are exactly the tools next on your roadmap
- **Native embedding model integrations:** Hugging Face, Google, OpenAI

## Common Use Cases

- Personalized recommender systems
- Document search engines (vector + full-text search combined)
- Image retrieval via text queries (multi-modal)
- Chatbots with semantic search/retrieval for context augmentation — **this is your RAG use case**
